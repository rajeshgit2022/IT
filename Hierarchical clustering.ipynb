# Dendrogram A dendrogram is the tree-like diagram used to visualize this process, plotting data points on one axis and the distance (dissimilarity) at which clusters are merged or split on the other

```python
X = df.iloc[:, [2,3]].values
```

```python
import scipy.cluster.hierarchy as sch
```

```python
#ward is like how "distance between clusters" is measured when deciding which two clusters to merge next
data = sch.linkage(X, method = 'ward')
```

```python
dendrogram = sch.dendrogram(data)
plt.figure(1,figsize=(15,15))

plt.title('Dendrogram')
plt.xlabel('customers')
plt.ylabel('eclidean distance')
plt.show()
```

# Hierarchical Clustering Model 

```python
#Agglomerative clustering is a popular "bottom-up" hierarchical clustering technique used in machine learning. It treats every data point as an individual cluster initially, and then iteratively merges the closest pairs of clusters together based on their similarity until all points are combined into a single, large cluste
from sklearn.cluster import AgglomerativeClustering
# affinity means Euclidean distance (straight-line distance) to measure how far apart points/clusters are
hc = AgglomerativeClustering(n_clusters = 5 , affinity = 'euclidean', linkage = 'ward')

y_hc = hc.fit_predict(X)
```

```python
y_hc
```

# visualization of scatterplot

```python
# plt.savefig('subplot.jpg')

plt.figure(1,figsize=(10,10))

plt.scatter(X[y_hc==0,0], X[y_hc ==0,1],s = 100 ,c = 'red', label = 'cluster0')

plt.scatter(X[y_hc==1,0], X[y_hc ==1,1],s = 100 ,c = 'blue', label = 'cluster1')

plt.scatter(X[y_hc==2,0], X[y_hc ==2,1],s = 100 ,c = 'green', label = 'cluster2')

plt.scatter(X[y_hc==3,0], X[y_hc ==3,1],s = 100 ,c = 'cyan', label = 'cluster3')

plt.scatter(X[y_hc==4,0], X[y_hc ==4,1],s = 100 ,c = 'magenta', label = 'cluster4')


plt.title('cluster of customers')
plt.xlabel('annual income')
plt.ylabel('spending score')
plt.legend()
plt.show()
```
