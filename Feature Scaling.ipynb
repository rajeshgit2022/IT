```python
#Encoding
df1 = pd.DataFrame({"shirtsize": ["x1","small", "medium", "large", "x1"]})
df1
```

```python
df1["shirtsize"].unique()
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder (categories = [["small","medium", "large", "x1"]])
df1["oe_shirtsize"] = oe.fit_transform (df1[["shirtsize"]]).astype('int')
df1
```
