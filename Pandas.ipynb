```python
 temperature_df = pd.DataFrame({
"city": ["mumbai","delhi","banglore", 'hyderabad'],
 "temperature": [32,45,4,36]})

temperature_df
```

```python
humidity_df = pd.DataFrame ({
 "city": ["delhi","mumbai", "banglore"],
 "humidity": [68, 65, 75]})
humidity_df
```

```python
                                                    #Combining Dataframes
#option 1: Append

#a Adding rows to the data frame by ading another data frame
df = temperature_df.append(humidity_df)
df
```

```python
temperature_df.append(humidity_df)
```

```python
#option 2: Concatenate
df = pd.concat([temperature_df, humidity_df])
df
```

```python
#if you want continuous index

df = pd.concat([temperature_df, humidity_df],ignore_index=True)
df
```

```python
df = pd.concat([temperature_df, humidity_df],axis=1)
df
```

```python
#Merging of DataFrame
#Inner Join
df = pd.merge(temperature_df, humidity_df, on='city',how="inner")
df
```

```python
#Outer Join
df = pd.merge(temperature_df, humidity_df, on='city', how="outer")
df
```
